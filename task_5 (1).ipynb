#Import Libraries

```python
import pandas as pd

from transformers import pipeline

from sklearn.metrics import accuracy_score
```

#Create Small Dataset

```python
data = {
    'ticket': [
        'Internet connection is very slow',
        'I want refund for my payment',
        'Unable to login into account',
        'Need help updating profile',
        'My bill is incorrect',
        'Password reset link not working',
        'Website is crashing frequently',
        'Need information about services'
    ],

    'label': [
        'Technical Support',
        'Billing',
        'Account Access',
        'General Inquiry',
        'Billing',
        'Account Access',
        'Technical Support',
        'General Inquiry'
    ]
}

df = pd.DataFrame(data)

df.head()
```

#Load Zero-Shot Model

```python
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)
```

#Define Labels

```python
candidate_labels = [
    "Technical Support",
    "Billing",
    "Account Access",
    "General Inquiry"
]
```

#Predict Tags

```python
predictions = []

for ticket in df['ticket']:

    result = classifier(
        ticket,
        candidate_labels
    )

    predictions.append(result['labels'][0])

predictions
```

#Evaluate Accuracy

```python
accuracy = accuracy_score(df['label'], predictions)

print("Accuracy Score:")
print(accuracy)
```

#Show Top 3 Tags

```python
for ticket in df['ticket']:

    result = classifier(
        ticket,
        candidate_labels
    )

    print("\nTicket:")
    print(ticket)

    print("\nTop 3 Predicted Tags:")

    for label, score in zip(result['labels'][:3], result['scores'][:3]):
        print(f"{label} : {score:.4f}")
```

#Few-Shot Prompting Example

```python
few_shot_example = """
Examples:

Ticket: Unable to login into account
Category: Account Access

Ticket: Need refund for my payment
Category: Billing

Ticket: Internet connection is slow
Category: Technical Support
"""

print(few_shot_example)
```
